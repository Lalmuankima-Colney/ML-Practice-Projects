```python
from nltk.corpus import words as nltk_words
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer, OrdinalEncoder
from sklearn.feature_extraction.text import TfidfVectorizer
import re
from sklearn.base import BaseEstimator, TransformerMixin


ENGLISH_WORDS = set(nltk_words.words())
STOPWORDS = set(stopwords.words('english'))
wnl = WordNetLemmatizer()
nlp = spacy.load('en_core_web_sm', disable=['parser', 'ner'])
categorical_cols = ['name', 'category', 'description', 'brand', 'specs']

class RandomSampleImputer(BaseEstimator, TransformerMixin):
    def __init__(self, random_state=None):
        self.random_state = random_state

    def fit(self, X, y=None):
        X = pd.DataFrame(X)
        self.fill_values_ = {col: X[col].dropna() for col in X.columns}
        self.feature_names_in_ = np.array(X.columns, dtype=object)
        return self

    def transform(self, X):
        X = pd.DataFrame(X).copy()
        for col in X.columns:
            n_missing = X[col].isnull().sum()
            if n_missing > 0:
                sampled = self.fill_values_[col].sample(
                    n_missing, replace=True, random_state=self.random_state
                ).values
                X.loc[X[col].isnull(), col] = sampled
        return X

    def get_feature_names_out(self, input_features=None):
        if input_features is not None:
            return np.array(input_features, dtype=object)
        return self.feature_names_in_


def lowercasing(A):
    A = pd.DataFrame(A).copy()
    for col in A.columns:
        A[col] = A[col].astype(str).str.lower()
    return A


impute_then_lower = Pipeline([
    ('impute', RandomSampleImputer(random_state=42)),
    ('lower', FunctionTransformer(lowercasing, feature_names_out="one-to-one")),
])

trf1 = ColumnTransformer(
    [
        ('impute_lower_23', impute_then_lower, categorical_cols),
    ],
    remainder='passthrough',
    verbose_feature_names_out=False,
)
trf1.set_output(transform="pandas")



class DescriptionWordFilter(BaseEstimator, TransformerMixin):
    def __init__(self, english_words, whitelist=None, min_len=3,
                 keep_alnum_specs=True, extra_vocab_from_data=0,
                 caesar_shift=3):
        self.english_words = english_words
        self.whitelist = whitelist or set()
        self.min_len = min_len
        self.keep_alnum_specs = keep_alnum_specs
        self.extra_vocab_from_data = extra_vocab_from_data
        self.caesar_shift = caesar_shift
        # matches tokens like '128GB', '15cm', 'XL', '4K' -- letters+digits mixed,
        # or short all-caps tokens (size/unit codes), not just plain words
        self._spec_pattern = re.compile(r'^(?=.*[0-9])[a-zA-Z0-9]+$|^[A-Za-z]{1,4}$')

    def _decode(self, text):
        if not self.caesar_shift or not isinstance(text, str):
            return text
        shift = self.caesar_shift % 26
        out = []
        for ch in text:
            if 'a' <= ch <= 'z':
                out.append(chr((ord(ch) - ord('a') - shift) % 26 + ord('a')))
            elif 'A' <= ch <= 'Z':
                out.append(chr((ord(ch) - ord('A') - shift) % 26 + ord('A')))
            else:
                out.append(ch)
        return ''.join(out)

    def fit(self, X, y=None):

        Xdf = pd.DataFrame(X)
        self.feature_name_in_ = Xdf.columns[0] if len(Xdf.columns) else 'text'
    
        self.whitelist_ = set(w.lower() for w in self.whitelist)
        if self.extra_vocab_from_data:
            # learn extra domain vocabulary from training text: decode first,
            # then count tokens that show up often but aren't in the
            # dictionary, instead of relying on hardcoded exceptions.
            from collections import Counter
            col = pd.DataFrame(X).iloc[:, 0].astype(str)
            counts = Counter()
            for text in col:
                decoded = self._decode(text)
                for tok in re.findall(r"[A-Za-z']+", decoded.lower()):
                    if tok not in self.english_words and len(tok) > self.min_len:
                        counts[tok] += 1
            learned = {tok for tok, c in counts.items()
                       if c >= self.extra_vocab_from_data}
            self.whitelist_ |= learned
        return self

    def _keep_token(self, tok):
        low = tok.lower()
        if low in self.english_words and len(tok) > self.min_len - 1:
            return True
        if low in self.whitelist_:
            return True
        if self.keep_alnum_specs and self._spec_pattern.match(tok):
            return True
        return False

    def _clean_text(self, text):
        if not isinstance(text, str) or not text.strip():
            return ''
        decoded = self._decode(text)
        tokens = re.findall(r"[A-Za-z0-9']+", decoded)  # strip stray punctuation while tokenizing
        kept = [tok for tok in tokens if self._keep_token(tok)]
        return ' '.join(kept)

    def transform(self, X):
        col = pd.DataFrame(X).iloc[:, 0]
        return col.apply(self._clean_text).to_frame(name=col.name or 'text')

    def get_feature_names_out(self, input_features=None):
        if input_features is not None:
            return np.array(input_features[:1], dtype=object)
        name = getattr(self, 'feature_name_in_', 'text')
        return np.array([name], dtype=object)


name_filter = DescriptionWordFilter(
    english_words=ENGLISH_WORDS,
    whitelist={'shapewear'},
    min_len=2,
    keep_alnum_specs=True,
    extra_vocab_from_data=5,
    caesar_shift=3,
)

description_filter = DescriptionWordFilter(
    english_words=ENGLISH_WORDS,
    whitelist={'shapewear'},
    min_len=2,
    keep_alnum_specs=True,
    extra_vocab_from_data=5,
    caesar_shift=3,
)

trf2 = ColumnTransformer(
    [
        ('cleaned_name', name_filter, ['name']),
        ('cleaned_description', description_filter, ['description']),
    ],
    remainder='passthrough',
    verbose_feature_names_out=False,
)
trf2.set_output(transform="pandas")




def _ruby_hash_to_py(raw_val):
    return raw_val.replace("=>", ":")


def _extract_pairs(specs):
    pairs = []
    if isinstance(specs, dict):
        specs = [specs]
    if isinstance(specs, list):
        for item in specs:
            if not isinstance(item, dict):
                continue
            k, v = item.get('key'), item.get('value')
            if k is not None or v is not None:
                pairs.append((k, v))
    return pairs


def dealing_with_specs(N):
    N = pd.DataFrame(N)
    col = N.iloc[:, 0]
    spec_keys_out, spec_values_out = [], []

    for raw_val in col:
        if not isinstance(raw_val, str):
            spec_keys_out.append(None)
            spec_values_out.append(None)
            continue
        try:
            val = ast.literal_eval(_ruby_hash_to_py(raw_val))
        except (ValueError, SyntaxError):
            spec_keys_out.append(None)
            spec_values_out.append(None)
            continue

        pairs = _extract_pairs(val.get('product_specification', {}))
        if not pairs:
            spec_keys_out.append(None)
            spec_values_out.append(None)
            continue

        # Join ALL keys/values into two parallel text blobs instead of
        # keeping only the first pair -- these then flow naturally into
        # the tfidf branch in trf6 as free text.
        keys = [str(k).strip().lower() for k, _ in pairs if k is not None]
        values = [str(v).strip().lower() for _, v in pairs if v is not None]
        spec_keys_out.append(' '.join(keys) if keys else None)
        spec_values_out.append(' '.join(values) if values else None)

    return pd.DataFrame(
        {'spec_key': spec_keys_out, 'spec_value': spec_values_out},
        index=col.index,
    )


def specs_feature_names(transformer, input_features):
    return np.array(['spec_key', 'spec_value'], dtype=object)


trf3 = ColumnTransformer(
    [
        ('separated', FunctionTransformer(dealing_with_specs, feature_names_out=specs_feature_names),
         ['specs']),
    ],
    remainder='passthrough',
    verbose_feature_names_out=False,
)
trf3.set_output(transform="pandas")





class WordStatsFeaturizer(BaseEstimator, TransformerMixin):
    def __init__(self, word_count_cols=None, pair_cols=None, use_union=False):
        self.word_count_cols = word_count_cols or ['category', 'description', 'spec_value']
        self.pair_cols = pair_cols or [
            ('category', 'spec_key'),
            ('category', 'spec_value'),
            ('spec_key', 'spec_value'),
        ]
        self.use_union = use_union

    def fit(self, X, y=None):
        return self  # stateless

    @staticmethod
    def _tokenize(text):
        if not isinstance(text, str) or not text.strip():
            return []
        tokens = [w.lower().strip(string.punctuation) for w in text.split()]
        return [w for w in tokens if w]

    def transform(self, X):
        X = pd.DataFrame(X).copy()
        new_cols = {}

        for col in self.word_count_cols:
            if col not in X.columns:
                raise KeyError(
                    f"WordStatsFeaturizer needs column '{col}' but it's missing. "
                    f"Available: {list(X.columns)}. If this runs after trf4, "
                    f"'category' no longer exists -- use 'Level_1' instead, or "
                    f"move this step earlier."
                )
            new_cols[f'{col}_num_words'] = X[col].apply(lambda t: len(self._tokenize(t)))

        for a, b in self.pair_cols:
            for col in (a, b):
                if col not in X.columns:
                    raise KeyError(f"Missing column '{col}' for pair ({a}, {b}).")
            wa = X[a].apply(lambda t: set(self._tokenize(t)))
            wb = X[b].apply(lambda t: set(self._tokenize(t)))

            common = pd.Series([len(x & y) for x, y in zip(wa, wb)], index=X.index)
            total = (
                pd.Series([len(x | y) for x, y in zip(wa, wb)], index=X.index)
                if self.use_union else
                pd.Series([len(x) + len(y) for x, y in zip(wa, wb)], index=X.index)
            )
            share = (common / total.replace(0, np.nan)).round(2).fillna(0.0)

            prefix = f'{a}__{b}'
            new_cols[f'{prefix}_word_common'] = common
            new_cols[f'{prefix}_word_total'] = total

        return pd.concat([X, pd.DataFrame(new_cols, index=X.index)], axis=1)

    def get_feature_names_out(self, input_features=None):
        names = [f'{c}_num_words' for c in self.word_count_cols]
        for a, b in self.pair_cols:
            names += [f'{a}__{b}_word_common', f'{a}__{b}_word_total']
        base = list(input_features) if input_features is not None else []
        return np.array(base + names, dtype=object)


trf3_new = WordStatsFeaturizer(
    word_count_cols=[],
    pair_cols=[
        ('spec_key', 'spec_value'),
    ],
)




class CategoryLevelSplitter(BaseEstimator, TransformerMixin):
    def __init__(self, sep=r'\s*>>\s*'):
        self.sep = sep

    def fit(self, X, y=None):
        X = pd.DataFrame(X)
        s = X.iloc[:, 0]
        split = s.str.split(self.sep, expand=True, regex=True)
        self.n_levels_ = split.shape[1]
        return self

    def transform(self, X):
        X = pd.DataFrame(X)
        s = X.iloc[:, 0]
        split = s.str.split(self.sep, expand=True, regex=True)

        # pad if this batch has fewer levels than training data saw
        for i in range(split.shape[1], self.n_levels_):
            split[i] = ''
        # truncate if this batch somehow has more
        split = split.iloc[:, :self.n_levels_]
        split = split.fillna('')  # guard against ragged rows producing NaN

        split.columns = [f"Level_{i + 1}" for i in range(self.n_levels_)]
        return split

    def get_feature_names_out(self, input_features=None):
        return np.array([f"Level_{i + 1}" for i in range(self.n_levels_)], dtype=object)


trf4 = ColumnTransformer(
    [
        ('fixed', CategoryLevelSplitter(), ['category']),
    ],
    remainder='passthrough',
    verbose_feature_names_out=False,
)
trf4.set_output(transform="pandas")




def remove_stopword(text):
    if not isinstance(text, str):
        return text
    kept = [w for w in text.split() if w.lower() not in STOPWORDS]
    return " ".join(kept)


def remove_punc(text):
    if not isinstance(text, str):
        return text
    return text.translate(str.maketrans('', '', string.punctuation))


def lemmatize_text(text):
    if not isinstance(text, str):
        return text
    tokens = nltk.word_tokenize(text)
    lemmas = [wnl.lemmatize(tok, pos='v') for tok in tokens]
    return " ".join(lemmas)


def spacy_lemmatize(text):
    if not isinstance(text, str):
        return text
    doc = nlp(text)
    return " ".join(tok.lemma_ for tok in doc)


def full_text_clean(N):
    N = pd.DataFrame(N)
    result = N.copy()
    for col in result.columns:
        result[col] = (
            result[col]
            .apply(remove_punc)
            .apply(remove_stopword)
            .apply(lemmatize_text)  # swap for spacy_lemmatize if preferred
        )
    return result


trf5 = ColumnTransformer(
    [
        ('lemmatized', FunctionTransformer(full_text_clean, feature_names_out="one-to-one"),
         ['description', 'brand', 'name']),
    ],
    remainder='passthrough',
    verbose_feature_names_out=False,
)
trf5.set_output(transform="pandas")






def select_level_cols(X):
    X = pd.DataFrame(X)
    return [c for c in X.columns if c.startswith('Level_')]


def clean_text_col(X):
    # X arrives as a 1D array/Series (single-column selection), not a DataFrame
    s = pd.Series(X).fillna('')
    return s.astype(str)


def make_field_tfidf(max_features):
    return Pipeline([
        ('fillna', FunctionTransformer(clean_text_col, validate=False)),
        ('tfidf', TfidfVectorizer(max_features=max_features, min_df=1)),
    ])


# trf6 = ColumnTransformer(
#     [
#         ('name_tfidf', make_field_tfidf(10000), 'name'),
#         ('desc_tfidf', make_field_tfidf(20000), 'description'),
#         ('brand_tfidf', make_field_tfidf(2000), 'brand'),
#         ('spec_key_tfidf', make_field_tfidf(3000), 'spec_key'),
#         ('spec_value_tfidf', make_field_tfidf(5000), 'spec_value'),
#         ('level_encode', OrdinalEncoder(
#             handle_unknown='use_encoded_value', unknown_value=-1
#         ), select_level_cols),
#     ],
#     # Only the numeric word-stat columns (spec_key__spec_value_word_common /
#     # _word_total) remain in remainder -- everything else is explicitly
#     # handled above, so passthrough never sees a raw string column.
#     remainder='passthrough',
#     verbose_feature_names_out=False,
# )


# TEXT EMBEDDING
from sentence_transformers import SentenceTransformer
import scipy.sparse as sp


class SentenceEmbeddingVectorizer(BaseEstimator, TransformerMixin):
    def __init__(self, model_name='all-MiniLM-L6-v2', batch_size=64,
                 device=None, normalize_embeddings=True):
        self.model_name = model_name
        self.batch_size = batch_size
        self.device = device
        self.normalize_embeddings = normalize_embeddings

    def _get_model(self):
        # cache per (model_name, device) on the instance; avoids reloading
        # the model on every transform() call within the same fitted pipe
        if not hasattr(self, '_model') or self._model_key != (self.model_name, self.device):
            self._model = SentenceTransformer(self.model_name, device=self.device)
            self._model_key = (self.model_name, self.device)
        return self._model

    def fit(self, X, y=None):
        s = clean_text_col(pd.DataFrame(X).iloc[:, 0])
        self.n_features_out_ = self._get_model().get_sentence_embedding_dimension()
        return self

    def transform(self, X):
        s = clean_text_col(pd.DataFrame(X).iloc[:, 0])
        model = self._get_model()
        emb = model.encode(
            s.tolist(),
            batch_size=self.batch_size,
            show_progress_bar=False,
            normalize_embeddings=self.normalize_embeddings,  # unit-norm -> dot product = cosine sim
            convert_to_numpy=True,
        )
        return sp.csr_matrix(emb)

    def get_feature_names_out(self, input_features=None):
        dim = getattr(self, 'n_features_out_', None) or self._get_model().get_sentence_embedding_dimension()
        return np.array([f'desc_emb_{i}' for i in range(dim)], dtype=object)


desc_embedder = SentenceEmbeddingVectorizer(
    model_name='all-MiniLM-L6-v2',   # 384-dim, fast, good general-purpose baseline
    batch_size=64,
    normalize_embeddings=True,
)

trf6 = ColumnTransformer(
    [
        ('name_tfidf', make_field_tfidf(10000), 'name'),
       #  ('desc_tfidf', make_field_tfidf(20000), 'description'),
        ('desc_embed', desc_embedder, 'description'),   # <-- swapped in
        ('brand_tfidf', make_field_tfidf(2000), 'brand'),
        ('spec_key_tfidf', make_field_tfidf(3000), 'spec_key'),
        ('spec_value_tfidf', make_field_tfidf(5000), 'spec_value'),
        ('level_encode', OrdinalEncoder(
            handle_unknown='use_encoded_value', unknown_value=-1
        ), select_level_cols),
    ],
    remainder='passthrough',
    verbose_feature_names_out=False,
    sparse_threshold=1.0,)

cat_pipe = Pipeline([
    ('trf1', trf1),
    ('trf2', trf2),
    ('trf3', trf3),
    ('trf3_new', trf3_new),
    ('trf4', trf4),
    ('trf5', trf5),
    ('trf6', trf6),
])


if __name__ == "__main__":
    # Example usage (uncomment and adapt):
    # raw_df = pd.read_csv("your_data.csv")[categorical_cols]
    # raw_df = raw_df.reset_index(drop=True)  # extra safety against index drift
    # Xt = cat_pipe.fit_transform(raw_df)
    pass
```
